```python
import lzma, tarfile
from pathlib import Path

db_dir = Path.cwd() / "reference_data" / "phononDB"          # folder the repository will live in
out = db_dir / "phonon_db"
out.mkdir(exist_ok=True)

if not any(out.iterdir()):
    for f in (db_dir / "phonon_db_tarred").glob("*.lzma"):
        with lzma.open(f) as fh, tarfile.open(fileobj=fh) as tf:
            tf.extractall(out)
    print("Extracted")
else:
    print("Already extracted")
```

Extracts the compressed phonon database so it can be read by `create_lookup_json.py`.
- Creates a `phonon_db` folder inside `phononDB` (if it doesn't already exist).
- Decompresses each of the archives in `phonon_db_tarred` into `phonon_db`. Each archive becomes one material folder containing its `POSCAR-unitcell`, `FORCE_SETS` and `phonon.yaml` files.
- Replicates the `extract_files_linux.sh` / `extract_files_mac.sh` scripts for windows machine
- If `phonon_db` already contains files, extraction is skipped, so re-running the notebook won't extract again.
